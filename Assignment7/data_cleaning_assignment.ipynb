## 1. Load the CSV dataset into a Pandas DataFrame

## 3. Handle missing values

```python
# Identify missing values per column
missing_counts = df.isnull().sum()
missing_pct = (df.isnull().mean() * 100).round(2)
missing_summary = pd.DataFrame({"missing_count": missing_counts, "missing_pct": missing_pct})
missing_summary[missing_summary["missing_count"] > 0].sort_values("missing_count", ascending=False)
```

```python
# Strategy:
# - Numeric columns (Price, Quantity, Discount): fill with the column median
# - Categorical columns (Ship_Mode, Customer_Name): fill with a placeholder / mode

df_clean = df.copy()

numeric_fill_cols = ["Price", "Quantity", "Discount"]
for col in numeric_fill_cols:
    median_val = df_clean[col].median()
    df_clean[col] = df_clean[col].fillna(median_val)
    print(f"Filled missing {col} with median: {median_val}")

categorical_fill_cols = ["Ship_Mode", "Customer_Name"]
for col in categorical_fill_cols:
    mode_val = df_clean[col].mode()[0]
    df_clean[col] = df_clean[col].fillna(mode_val)
    print(f"Filled missing {col} with mode: {mode_val}")
```

```python
# Confirm no missing values remain in the columns we handled
df_clean[numeric_fill_cols + categorical_fill_cols].isnull().sum()

# If any row still has missing critical identifiers (e.g., Order_ID), it would be safer to drop it.
# Example (not needed here since Order_ID has no missing values):
# df_clean = df_clean.dropna(subset=["Order_ID"])
df_clean.isnull().sum().sum()  # total remaining missing values
```

## 4. Basic operations — filter rows & select columns

```python
# Select a subset of columns
subset_cols = ["Order_ID", "Category", "Sub_Category", "Region", "Price", "Quantity"]
df_clean[subset_cols].head()
```

```python
# Filter rows: Technology orders with Price > 200
tech_high_value = df_clean[(df_clean["Category"] == "Technology") & (df_clean["Price"] > 200)]
print(f"{len(tech_high_value)} Technology orders with Price > 200")
tech_high_value[subset_cols].head()
```

```python
# Filter rows: orders shipped via Same Day
same_day = df_clean[df_clean["Ship_Mode"] == "Same Day"]
print(f"{len(same_day)} orders shipped Same Day")
same_day[["Order_ID", "Ship_Mode", "Region"]].head()
```

## 5. Remove duplicates

```python
# Check for exact duplicate rows
dup_count = df_clean.duplicated().sum()
print(f"Duplicate rows found: {dup_count}")
df_clean[df_clean.duplicated(keep=False)].sort_values("Order_ID").head()
```

```python
# Drop duplicates, keeping the first occurrence
before = len(df_clean)
df_clean = df_clean.drop_duplicates(keep="first").reset_index(drop=True)
after = len(df_clean)
print(f"Removed {before - after} duplicate rows. New shape: {df_clean.shape}")
```

## 6. Create a derived column: `total_amount = price * quantity`

```python
df_clean["total_amount"] = (df_clean["Price"] * df_clean["Quantity"]).round(2)
df_clean[["Order_ID", "Price", "Quantity", "total_amount"]].head()
```

```python
# Sanity check on the new column
df_clean["total_amount"].describe()
```

## 7. Save the cleaned dataset as a new CSV

```python
OUTPUT_PATH = "../data/superstore_cleaned.csv"
df_clean.to_csv(OUTPUT_PATH, index=False)
print(f"Saved cleaned dataset to {OUTPUT_PATH}")
print(f"Final shape: {df_clean.shape}")
```

- **Filtering**: demonstrated row filtering (e.g., high-value Technology orders, Same Day shipments) and
